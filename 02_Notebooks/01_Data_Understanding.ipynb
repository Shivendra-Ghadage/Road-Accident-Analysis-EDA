## 1. Business Problem

Road collisions are influenced by multiple factors such as road conditions,
weather, lighting, speed limits, road type, junction characteristics, and
the time at which a collision occurs.

Analyzing these factors can help identify patterns associated with accident
frequency and severity.

The purpose of this project is to explore road collision data and identify
important patterns, relationships, and risk factors that can support
data-driven road safety analysis.
